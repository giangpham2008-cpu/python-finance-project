```python
import matplotlib.pyplot as plt
import matplotlib.dates as dates

fig, ax = plt.subplots(figsize=(10, 5))

def plot_candle(df, ax):
    # Râu nến
    ax.vlines(
        x=df.index, ymin=df.Low, ymax=df.High,
        colors='k', linewidth=1
    )

    # Nến đỏ: giá giảm
    red = df.query('Open > Close')
    ax.vlines(
        x=red.index, ymin=red.Close, ymax=red.Open,
        colors='r', linewidth=3
    )

    # Nến xanh: giá tăng hoặc không đổi
    green = df.query('Open <= Close')
    ax.vlines(
        x=green.index, ymin=green.Open, ymax=green.Close,
        colors='g', linewidth=3
    )

    ax.xaxis.set_major_locator(dates.MonthLocator())
    ax.xaxis.set_major_formatter(dates.DateFormatter('%b-%y'))
    ax.xaxis.set_minor_locator(dates.DayLocator())

    return df

(raw
    .iloc[:, ::2]
    .pipe(fix_cols)
    .resample('D')
    .agg({
        'Open': 'first',
        'High': 'max',
        'Low': 'min',
        'Close': 'last'
    })
    .loc['2019-09':'2019-12']
    .pipe(plot_candle, ax)
)
```

```python
def calc_obv(df):
    df = df.copy()
    df["OBV"] = 0.0
    obv_col = df.columns.get_loc("OBV")

    for i in range(1, len(df)):
        close_today = df["Close"].iloc[i]
        close_prev = df["Close"].iloc[i - 1]
        volume_today = df["Volume"].iloc[i]
        obv_prev = df["OBV"].iloc[i - 1]

        if close_today > close_prev:
            df.iat[i, obv_col] = obv_prev + volume_today
        elif close_today < close_prev:
            df.iat[i, obv_col] = obv_prev - volume_today
        else:
            df.iat[i, obv_col] = obv_prev

    return df


# Tính OBV theo thứ tự ngày từ cũ đến mới
result = calc_obv(aapl.sort_index())

# Thêm giá phiên trước để dễ đối chiếu
result["Close_prev"] = result["Close"].shift(1)

# Hiện 10 dòng đầu
result[["Close_prev", "Close", "Volume", "OBV"]].head(10)
```

```python
def calc_ad(df, close_col='Close', low_col='Low',
            high_col='High', vol_col='Volume'):
    close = df[close_col]
    low = df[low_col]
    high = df[high_col]

    return (
        df.assign(
            mfm=((close - low) - (high - close)) / (high - low),
            mfv=lambda df_: df_.mfm * df_[vol_col],
            cmfv=lambda df_: df_.mfv.cumsum()
        )
        .cmfv
    )
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# Tính trung bình theo công thức của thầy
def avg(df, col, window_size=14):
    results = []
    window = []

    for i, val in enumerate(df[col]):
        window.append(val)

        if i < window_size:
            results.append(np.nan)

        elif i == window_size:
            window.pop(0)
            results.append(sum(window) / window_size)

        else:
            results.append(
                (results[-1] * (window_size - 1) + val)
                / window_size
            )

    return pd.Series(results, index=df.index)


# Dùng dữ liệu APPLE
rsi_data = (
    aapl[['Close']]
    .assign(
        change=lambda df: df['Close'].diff(),
        gain=lambda df: df.change.clip(lower=0),
        loss=lambda df: df.change.clip(upper=0),
        avg_gain=lambda df: avg(df, 'gain'),
        avg_loss=lambda df: -avg(df, 'loss'),
        rs=lambda df: df.avg_gain / df.avg_loss,
        rsi=lambda df: np.select(
            [df.avg_loss == 0],
            [100],
            default=100 - 100 / (1 + df.rs)
        )
    )
)

# Hiện bảng kết quả
display(rsi_data)

# Vẽ biểu đồ RSI của APPLE
ax = rsi_data['rsi'].plot(
    figsize=(12, 4),
    title='Apple — RSI (14)',
    color='purple'
)
ax.axhline(70, color='red', linestyle='--', label='70')
ax.axhline(30, color='green', linestyle='--', label='30')
ax.set_ylim(0, 100)
ax.set_ylabel('RSI')
ax.legend()
plt.show()
```

```python
import numpy as np
import numba as nb
import matplotlib.pyplot as plt
from IPython.display import display

@nb.njit
def avg_nb(ser, window_size=14):
    results = np.full(len(ser), np.nan)

    if len(ser) <= window_size:
        return results

    # Trung bình đầu tiên: 14 mức thay đổi
    total = 0.0
    for i in range(1, window_size + 1):
        total += ser[i]
    results[window_size] = total / window_size

    # Các trung bình tiếp theo
    for i in range(window_size + 1, len(ser)):
        results[i] = (
            results[i - 1] * (window_size - 1) + ser[i]
        ) / window_size

    return results

apple_rsi_nb = (
    aapl[["Close"]]
    .assign(
        change=lambda df: df["Close"].diff(),
        gain=lambda df: df.change.clip(lower=0),
        loss=lambda df: df.change.clip(upper=0),
        avg_gain=lambda df: avg_nb(
            df.gain.to_numpy(dtype=np.float64), 14
        ),
        avg_loss=lambda df: -avg_nb(
            df.loss.to_numpy(dtype=np.float64), 14
        ),
        rs=lambda df: df.avg_gain / df.avg_loss,
        rsi=lambda df: np.where(
            df.avg_loss == 0,
            100,
            100 - 100 / (1 + df.rs)
        )
    )
)

display(apple_rsi_nb.tail(20))

ax = apple_rsi_nb.rsi.plot(
    figsize=(12, 4), title="Apple — RSI (14), Numba"
)
ax.axhline(70, color="red", linestyle="--")
ax.axhline(30, color="green", linestyle="--")
ax.set_ylim(0, 100)
plt.show()
```

```python
import pandas as pd
from time import perf_counter

big = np.random.default_rng(42).random(10_000_000)
big_df = pd.DataFrame({"val": big})

# Biên dịch trước để không tính thời gian biên dịch vào phép đo
avg_nb(big[:100], 14)

start = perf_counter()
fast = avg_nb(big, 14)
fast_seconds = perf_counter() - start

start = perf_counter()
basic = avg(big_df, "val", 14)
basic_seconds = perf_counter() - start

np.testing.assert_allclose(
    fast, basic.to_numpy(), equal_nan=True
)

print("Hai cách cho kết quả khớp nhau.")
print(f"Bản thường: {basic_seconds:.3f} giây")
print(f"Bản Numba:  {fast_seconds:.3f} giây")
print(f"Tỷ lệ thời gian thường / Numba: {basic_seconds / fast_seconds:.2f}")
```
